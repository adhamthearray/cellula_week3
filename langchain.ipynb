```python
import os
from dotenv import load_dotenv


load_dotenv()
token = os.getenv("llm_key")

```

```python
from langchain_openai import ChatOpenAI
llm=ChatOpenAI(
    api_key=token,
    base_url="https://openrouter.ai/api/v1",
    model="openrouter/free",
    temperature=0.8
)
```

```python

## step 1 : create vector database
# Loading
# Chunking
# Embedding
# Vector stores
# Loading
from langchain_classic.document_loaders import TextLoader
loader = TextLoader("data.txt")
document = loader.load()
# Chunking
from langchain_classic.text_splitter import TokenTextSplitter
splitter = TokenTextSplitter(
chunk_size = 200,
chunk_overlap = 20
)
chunks = splitter.split_documents(document)
# Embedding
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(
model_name = "sentence-transformers/all-MiniLM-L6-v2",
model_kwargs = {"device":"cpu"}
)
vectors = embeddings.embed_documents([i.page_content for i in chunks])
# vector store
from langchain_classic.vectorstores import FAISS
vectorDB = FAISS.from_documents(chunks,embeddings)
```

```python
query = "When did damien chazzle work with ryan"
similar_docs = vectorDB.similarity_search_with_score(query,k=3)
for i in similar_docs :
 print(i[0].page_content)
```

```python
from langchain_core.prompts import PromptTemplate
temp = """
you are ryan gosling, Answer the next Question using provided context,
If you don't know the answer, just say you don't know.
answer should be within 200 words or lower only
## context :
{context}
## Question :
{Question}
"""
temp = PromptTemplate.from_template(temp)
query = "what are Movies u did that had a huge success"

similar_docs = vectorDB.similarity_search_with_score(query,k=4)
context = []
for i in similar_docs :
    context.append(i[0].page_content)
prompt = temp.format(context="\n".join(context), Question = query)
print(prompt)
```

```python
response = llm.invoke(prompt).content
```
